Connecting ADLS GEN2 to Azure Databricks:

Your course instructor's code uses Credential Passthrough (spark.databricks.passthrough.adls.gen2.tokenProviderClassName) combined with legacy dbutils.fs.mount. Both of these are deprecated in Databricks and are blocked on Serverless workspaces (as well as newer Databricks Runtimes), which is why Spark throws CONFIG_NOT_AVAILABLE.
Instead of mounting /mnt/bronze-fs, connect directly to your storage account (adlsgen2adeproject1) using Unity Catalog External Locations (the modern industry standard that works cleanly on Serverless workspaces) and read/write using direct abfss:// paths.

```python
# 1. List folders inside your bronze container
# display(dbutils.fs.ls("abfss://bronze-fs@adlsgen2adeproject1.dfs.core.windows.net/"))
display(dbutils.fs.ls("abfss://bronze-fs@adlsgen2adeproject1.dfs.core.windows.net/SalesLT/"))
```

```python
# 2. Read one of your copied Parquet tables (example: Customer)
df = spark.read.parquet("abfss://bronze-fs@adlsgen2adeproject1.dfs.core.windows.net/SalesLT/Customer/")
display(df)
```
